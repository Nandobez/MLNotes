Como em qualquer outro exercício de programação, o primeiro passo é importar as bibliotecas necessárias. Como vamos usar o Google Colab para programar nossa rede neural, precisamos instalar e importar as bibliotecas do PyTorch.


```python
## The usual imports
import torch
import torch.nn as nn

## print out the pytorch version used
print(torch.__version__)
```

## Os Dados


Antes de construir e treinar a rede neural, o primeiro passo é processar e preparar os dados. Neste notebook vamos usar dados sintéticos (i.e., dados fictícios), então não vamos usar dados do mundo real.

Por simplicidade, vamos usar os pares de entrada e saída a seguir, convertidos em tensores, que é como os dados são tipicamente representados no mundo do deep learning. Os valores de `x` representam a entrada de dimensão `(6, 1)` e os valores de `y` representam a saída de dimensão similar.

O objetivo da rede neural que vamos construir e treinar é aprender automaticamente os padrões que melhor caracterizam a relação entre os valores de `x` e `y`. Essencialmente, o modelo aprende a relação que existe entre entradas e saídas, e essa relação pode ser usada depois para prever o valor de `y` correspondente a qualquer entrada `x`.


```python
## our data in tensor form
x = torch.tensor([[-1.0],  [0.0], [1.0], [2.0], [3.0], [4.0]], dtype=torch.float)
y = torch.tensor([[-3.0], [-1.0], [1.0], [3.0], [5.0], [7.0]], dtype=torch.float)
```

```python
## print size of the input tensor
x.size()
```

## Componentes da Rede Neural

Como dito antes, vamos primeiro definir e construir os componentes da nossa rede neural antes de treinar o modelo.

### Modelo

Tipicamente, ao construir uma rede neural definimos camadas e pesos que formam os componentes básicos do modelo. Abaixo mostramos um exemplo de como definir uma camada oculta chamada `layer1` com tamanho `(1, 1)`. Para fins deste tutorial, não vamos definir explicitamente os `pesos` — deixamos as funções nativas do PyTorch cuidarem disso. A propósito, a função `nn.Linear(...)` aplica uma transformação linear ($y = xA^T + b$) aos dados passados como entrada. Por enquanto ignoramos o bias com `bias=False`.


```python
## Neural network with 1 hidden layer
layer1 = nn.Linear(1,1, bias=False)
model = nn.Sequential(layer1)
```

### Perda e Otimizador

A função de perda, `nn.MSELoss()`, é responsável por informar ao modelo o quão bem ele aprendeu a relação entre entrada e saída. O otimizador (neste caso, `SGD`) tem como papel principal minimizar (ou reduzir) o valor da perda à medida que ajusta os pesos.


```python
## loss function
criterion = nn.MSELoss()

## optimizer algorithm
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

## Treinando a Rede Neural

Temos todos os componentes necessários para treinar o modelo. Abaixo está o código usado para treiná-lo.


```python
## training
for ITER in range(150):
    model = model.train()

    ## forward
    output = model(x)
    loss = criterion(output, y)
    optimizer.zero_grad()

    ## backward + update model params 
    loss.backward()
    optimizer.step()

    model.eval()
    print('Epoch: %d | Loss: %.4f' %(ITER, loss.detach().item()))
```

## Testando o Modelo

Depois de treinar o modelo, podemos testar a sua capacidade de predição passando uma entrada nova. Abaixo está um exemplo simples de como fazer isso com nosso modelo.


```python
## test the model
sample = torch.tensor([10.0], dtype=torch.float)
predicted = model(sample)
print(predicted.detach().item())
```
