## Intuição

Receitas modernas raramente mantêm o learning rate fixo: aquecem linearmente e depois decaem (cosseno, exponencial, em escada). Schedules impactam bastante a acurácia final.


## Formulação Matemática

Cosine com warmup de $W$ passos e total $T$:

$$\eta_t = \eta_{\max} \cdot \min\!\left(\frac{t}{W},\ \tfrac{1}{2}\big(1 + \cos(\pi \tfrac{t - W}{T - W})\big)\right)$$


## Implementação


```python
import math
import matplotlib.pyplot as plt
```

```python
def warmup_cosine(t, total, warmup, eta_max=1.0):
    if t < warmup:
        return eta_max * t / warmup
    p = (t - warmup) / (total - warmup)
    return 0.5 * eta_max * (1 + math.cos(math.pi * p))

def one_cycle(t, total, eta_max=1.0):
    half = total / 2
    if t < half:
        return eta_max * t / half
    return eta_max * (1 - (t - half) / half)
```

## Experimento


```python
total = 1000
ts = list(range(total))
plt.plot(ts, [warmup_cosine(t, total, warmup=100) for t in ts], label='warmup + cosine')
plt.plot(ts, [one_cycle(t, total) for t in ts], label='1cycle')
plt.legend(); plt.xlabel('step'); plt.ylabel('lr / max'); plt.title('LR schedules'); plt.show()
```

## Discussão

- Warmup é crítico ao usar AdamW com batches grandes — sem ele o second moment fica ruidoso.
- 1cycle (pico alto + decaimento simétrico) generaliza muito bem em datasets pequenos.
- Para pretraining/finetuning de LLMs o padrão é warmup + cosine com $\eta_{\min} \approx 0.1\,\eta_{\max}$.
