### Funções

Funções mapeiam um objeto matemático em outro. Em ML é importante entender o conceito — usamos funções o tempo todo. De fato, vários conceitos de ML são funções amarradas mapeando entradas a saídas. Vamos começar pelo básico e evoluir até funções mais comuns em ML.


Exemplo de função quadrática:

$$
f(x) = x^2
$$


```python
# import your main libraries

import matplotlib.pyplot as plt
import numpy as np
```

```python
# simple squaring function
def f(x):
    return x**2

# test the function
x = 2
print(f(x))
```

Vamos visualizar como fica:


```python
# visualize using matplotlib

x = np.linspace(-10, 10, 100)
y = f(x)
plt.plot(x, y)
```

Função constante:

$$
f(x) = c
$$

com $c$ constante.


```python
# a constant function
def f(x):
    return 2 # constant

# test the function
x = 2

# visualize using matplotlib
x = np.linspace(-10, 10, 100)
y = [f(x_i) for x_i in x]

plt.plot(x, y)
```

Função linear:

$$
f(x) = mx + c
$$

com $m$ a inclinação e $c$ o intercepto em $y$ — ambos constantes.


```python
# a linear function
def f(x, m, c):
    return m*x + c

# test the function
x = 2
m = 3
c = 4

# visualize using matplotlib
x = np.linspace(-10, 10, 100)
y = [f(x_i, m, c) for x_i in x]

plt.plot(x, y)
```

Funções polinomiais:

$$
f(x) = a_0 + a_1 x + a_2 x^2 + \cdots + a_n x^n
$$

com $a_0, a_1, \dots, a_n$ constantes e $n$ o grau do polinômio.


```python
# a polynomial function
def f(x, a, b, c):
    # a, b, c are coefficients
    return a*x**2 + b*x + c

# test the function
x = 2
a = 8
b = 1
c = 8

# visualize using matplotlib
x = np.linspace(-10, 10, 100)
y = [f(x_i, a, b, c) for x_i in x]

plt.plot(x, y)
```

Funções logarítmicas:

$$
f(x) = c\,\log_a x
$$

com $c$ e $a$ constantes e $\log_a$ o logaritmo na base $a$.


```python
# a logarithmic function
def f(x, a):
    return a*np.log(x)  # natural log

# test the function
x = 2
a = 3

# visualize using matplotlib
x = np.linspace(0.1, 10, 100)
y = [f(x_i, a) for x_i in x]

plt.plot(x, y)
```

Funções exponenciais:

$$
f(x) = c \cdot a^x
$$

com $c$ e $a$ constantes.


```python
# an exponential function
def f(x, a, c):
    return c*np.exp(a*x)

# test the function
x = 2
a = 3
c = 4

# visualize using matplotlib
x = np.linspace(-10, 10, 100)
y = [f(x_i, a, c) for x_i in x]

plt.plot(x, y)
```

### Raízes de Funções


Checar se uma função tem raiz. Abaixo verificamos uma função linear cuja raiz única é $x = -c/m$.


```python
# function to find root of a linear function

def f(x, m, c):
    return m*x + c

def has_root(x, m, c):
    return x == -c/m

# generate range of values that includes root of linear function
x = np.linspace(-10, 10, 100)

# add root to the list
x = np.append(x, -c/m)

# order the list of values
x = np.sort(x)

y = [has_root(x_i, m, c) for x_i in x]

# y to visualize
y = [f(x_i, m, c) for x_i in x]

# visualize using matplotlib, with red dots at roots
plt.plot(-c/m, f(-c/m, m, c), 'r.')
plt.plot(x, y)

# label root value with text x = -c/m
plt.text(-c/m, f(-c/m, m, c), 'x = -c/m'+str(-c/m))
```

### Transformações de Funções

Transformações são quando pegamos a saída de uma função e passamos por outra.


Translações: mover a função pra cima, baixo, esquerda ou direita. Exemplo de translação vertical:


```python
# function f(x) = x**2
def f(x):
    return x**2

# test the function
x = np.linspace(-4, 4, 100)

# visualize using matplotlib
plt.plot(x, f(x))

# plot function shifted by 1 unit vertically; use red
plt.plot(x, f(x)+1, 'r')
```

Translação horizontal:


```python
# plot function shifted by 2 units horizontally; use green
x = np.linspace(-5, 5, 100)
plt.plot(x, x**2, label='$f(x)=x^2$')
plt.plot(x, (x-2)**2, label='$f(x)=(x-2)^2$')

plt.xlim(-6, 6)
plt.ylim(-2, 4)

plt.plot(x, x*0, 'k', alpha=0.2)
plt.plot(x*0, x, 'k', alpha=0.2)
plt.legend()
```

Exemplo de escalonamento:


```python
# vertical scaling
x = np.linspace(-6, 6, 100)
plt.plot(x, x**2)
plt.plot(x, (x**2)/2, 'g')

plt.xlim(-6, 6)
plt.ylim(-2, 4)

plt.plot(x, x*0, 'k', alpha=0.2)
plt.plot(x*0, x, 'k', alpha=0.2)
```

Escalonamento por constante negativa.


```python
# horizontal scaling
x = np.linspace(-6, 6, 100)
plt.plot(x, x**2)
plt.plot(x, -(x**2)/2, 'g')

plt.xlim(-6, 6)
plt.ylim(-4, 4)

plt.plot(x, x*0, 'k', alpha=0.2)
plt.plot(x*0, x, 'k', alpha=0.2)

plt.plot(x, x**2, label='$f(x)=x^2$')
plt.plot(x, -(x**2)/2, 'g', label='$f(x)=-\\frac{1}{2}x^2$')
plt.legend()
```

Plot da função seno.


```python
x = np.linspace(-6, 6, 100)
plt.plot(x, np.sin(x), label='$f(x)=sin(x)$')

plt.plot(x, x*0, 'k', alpha=0.2)
plt.plot(x*0, x, 'k', alpha=0.2)

plt.legend()
```

Outro exemplo de translação.


```python
x = np.linspace(-6, 6, 100)
plt.plot(x, x**3 - x, label='$f(x)=x^3-x$')

# horizontal shift
plt.plot(x, (x+2)**3 - (x+2), 'g', label='$f(x)=(x+2)^3-(x+2)$')

plt.plot(x, x*0, 'k', alpha=0.2)
plt.plot(x*0, x, 'k', alpha=0.2)

plt.ylim(-5, 5)
plt.legend()
```

Escalonamento vertical do seno:


```python
x = np.linspace(-6, 6, 100)

# sine with vertical scaling
plt.plot(x, np.sin(x), label='$f(x)=sin(x)$')
plt.plot(x, 2*np.sin(x), 'g', label='$f(x)=2sin(x)$')

plt.plot(x, x*0, 'k', alpha=0.2)
plt.plot(x*0, x, 'k', alpha=0.2)

plt.ylim(-5, 5)
plt.legend()
```
