## Intuition

Integrated Gradients attributes a model's prediction to its inputs by integrating gradients along a straight line from a baseline (e.g. all zeros) to the actual input. The attribution sums to the *difference* in output, satisfying the *completeness* axiom.


## Mathematical Formulation

$$\text{IG}_i(x) = (x_i - x_i^\text{base}) \int_0^1 \frac{\partial f(x^\text{base} + \alpha(x - x^\text{base}))}{\partial x_i}\,d\alpha$$


## Implementation


```python
# pip install captum
import torch
import torch.nn as nn
from captum.attr import IntegratedGradients
```

```python
class TinyClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(4, 16), nn.ReLU(), nn.Linear(16, 2))
    def forward(self, x): return self.net(x)

torch.manual_seed(0)
model = TinyClassifier().eval()
```

## Experiment


```python
x = torch.tensor([[0.5, -1.0, 2.0, 0.0]])
baseline = torch.zeros_like(x)
ig = IntegratedGradients(model)
attr = ig.attribute(x, baselines=baseline, target=1, n_steps=64)
print('attribution per feature:', attr.squeeze().tolist())
print('sum of attributions  :', attr.sum().item())
print('output diff          :', (model(x)[0,1] - model(baseline)[0,1]).item())
```

## Discussion

- The choice of baseline matters: zero, average input, or a noisy baseline give different signals.
- Use `n_steps=50`–`200`; more steps shrink discretisation error but cost compute.
- Captum integrates with anything PyTorch — vision models, transformers (via embedding-layer attribution), tabular.
