## Intuition

Most modern recipes don't keep the learning rate fixed: they warm it up linearly, then decay (cosine, exponential, step). Schedules are surprisingly impactful on final accuracy.


## Mathematical Formulation

Cosine with warmup over $W$ warm-up steps and total $T$:

$$\eta_t = \eta_{\max} \cdot \min\!\left(\frac{t}{W},\ \tfrac{1}{2}\big(1 + \cos(\pi \tfrac{t - W}{T - W})\big)\right)$$


## Implementation


```python
import math
import matplotlib.pyplot as plt
```

```python
def warmup_cosine(t, total, warmup, eta_max=1.0):
    if t < warmup:
        return eta_max * t / warmup
    p = (t - warmup) / (total - warmup)
    return 0.5 * eta_max * (1 + math.cos(math.pi * p))

def one_cycle(t, total, eta_max=1.0):
    half = total / 2
    if t < half:
        return eta_max * t / half
    return eta_max * (1 - (t - half) / half)
```

## Experiment


```python
total = 1000
ts = list(range(total))
plt.plot(ts, [warmup_cosine(t, total, warmup=100) for t in ts], label='warmup + cosine')
plt.plot(ts, [one_cycle(t, total) for t in ts], label='1cycle')
plt.legend(); plt.xlabel('step'); plt.ylabel('lr / max'); plt.title('LR schedules'); plt.show()
```

## Discussion

- Warmup is critical when using AdamW with large batches — without it the second-moment estimate is noisy.
- 1cycle (high peak LR + symmetric decay) generalises great on small datasets.
- For pretraining/finetuning of LLMs the standard is warmup + cosine with $\eta_{\min} \approx 0.1\,\eta_{\max}$.
